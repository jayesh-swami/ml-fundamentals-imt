## Feature selection using Scikit-learn

Scikit-learn proposes several methods of *feature selection* using three approaches mentioned in the course (filter, *wrapper* and embedded). We will see here examples about filter and wrapper methods. [see documentation](https://scikit-learn.org/stable/modules/feature_selection.html#feature-selection).


### Filter approach
These two methods follow the **filter approach**, as the feature selection : the selection is made prior to the construction of a decision model.
 
- [Removing features with low variance](https://scikit-learn.org/stable/modules/feature_selection.html#removing-features-with-low-variance)  consiste of eliminating variables with a variance lower than a threshold (by default, the threshod is 0 and consequently constant variables are eliminated). As we have seen in the PCA course, a low variance variable is not necessary useless.  

- The [univariate feature selection](https://scikit-learn.org/stable/modules/feature_selection.html#univariate-feature-selection)  seeks to determine (via statistical tests)  to what extent each input feature explains the output feature (target variable).The variables with the lowest individual explanatory power are eliminated. 


### Wrapper approach
- The [sequential feature selection](https://scikit-learn.org/stable/modules/feature_selection.html#sequential-feature-selection) is a *wrapper* approach:   a decision model must be developed for each candidate (sub)set of input features and then the performance of the model will characterise and determine the (sub)set of features. Two versions are proposed, one incremental (*Forward-SFS*) and one decremental (*Backward-SFS*).



- [Recursive feature elimination](https://scikit-learn.org/stable/modules/feature_selection.html#recursive-feature-elimination) and [Feature selection using SelectFromModel](https://scikit-learn.org/stable/modules/feature_selection.html#feature-selection-using-selectfrommodel) : both methods rely on a decision model learned using all the input variables to provide indications of the relevance of each input variable; not all decision models provide such information. These methods can be seen as following the *wrapper* approach because (at least) a decision model must be developed to enable selection, but selection is then made on the basis of the "relevance" coefficients assigned (by the decision model) to the individual features. The cost of these two methods is therefore lower than the cost of generic *wrapper* methods, which have to learn a decision model for any candidate combination of variables.

## The dataset
To test some of the feature selection methods implemented in Scikit-learn we will use the Texture data (5500 observations which are pixels extracted [from texture images](https://www.robots.ox.ac.uk/~vgg/data/dtd/) and described by 40 variables which are local visual descriptors; each observation belongs to one of the 11 texture classes.

The last column contain the labels of the class : it is the target variable

```python
## here we use pandas juste to show clearly the content of the dataset file
import pandas as pd
engine='python'
df = pd.read_csv('texture.dat',  sep='\\s+|\\s+', header=None)
df
```

```python
### as the data are well represented by a matrix, we will use numpy package
## numpy contains a lot of useful function dedicated for handling matrix
import numpy as np
textures = np.loadtxt('texture.dat')
print(textures.shape)
textures
```

## Adding noise 
To illustrate the point, after reading the data, we add four input variables which represent noise following a [normal distribution](https://www.bibmath.net/dico/index.php?action=affiche&quoi=./l/loinormale.html) (for the first two) and a [uniform distribution](https://www.logamaths.fr/variable-aleatoire-continue-loi-uniforme/#) (for the last two).

### <font color='red'>The objectif of the tutorial is to use the appropriate methodes in order to eliminate these four random/noise variables</font>
    
    
We limit the observations to 110 (instead of 5500), chosen at random.

> 

```python

from sklearn.utils import shuffle
# generation of two normal feature and two uniform features
np.random.seed(1)
n_samples = 110
varUniformes = np.random.uniform(0, 1, size=(n_samples, 2))
varNormales = np.random.randn(n_samples, 2)
# adding the four new variables to the data infront of the rest of variables
donnees = np.hstack((varUniformes, varNormales, shuffle(textures, random_state=1, n_samples=n_samples)))
print(donnees.shape)
```

## Building the decision model
Recall that the last column contains the labels of the class : the target variable 

As a decision model, we choose to apply the [LinearDiscriminantAnalysis](https://scikit-learn.org/stable/modules/generated/sklearn.discriminant_analysis.LinearDiscriminantAnalysis.html)  : A classifier with a linear decision boundary, generated by fitting class conditional densities to the data and using Bayes’ rule.


> 

```python
from sklearn.model_selection import train_test_split
train, test = train_test_split(donnees, test_size = 0.5, random_state=1)
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
# sur données initiales
lda = LinearDiscriminantAnalysis()
```

We apply the LinearDiscriminantAnalysis first on the initial dataset and then on the dataset after adding the four new variables (noise).. 

Let us observe the output score : the mean accuracy on the given test data and labels

> 

```python
## on initial data
lda.fit(train[:,4:44],train[:,44])
lda.score(test[:,4:44],test[:,44])
#0.7272727272727273
```

```python
## on dataset after adding the four new variables
# sur données avec variables ajoutées
lda.fit(train[:,:44],train[:,44])
lda.score(test[:,:44],test[:,44])
#0.6909090909090909
```

It is easy to see that the presence of the four new variables, with random values, has a negative impact on the score and generalisability of the decision model constructed.

## Removing features with low variance

The first method that we will apply is [removing features with low variance](https://scikit-learn.org/stable/modules/feature_selection.html#removing-features-with-low-variance). 


To define the variance threshold, let us first apply the method with a default threshold equal to 0 (only constant variables will be eliminated):

> 

```python
from sklearn.feature_selection import VarianceThreshold
selVarianceThreshold = VarianceThreshold()
Xtrain_removed0variance = selVarianceThreshold.fit_transform(train[:,:44])
# show the variance of all variables
selVarianceThreshold.variances_
```

```python
# show of variances sorted in ascending order to define the threshold 
np.sort(selVarianceThreshold.variances_)
```

```python
## we can choose the threshold 0.04384, that allow to eliminate half of the variables
selVarianceThreshold2 = VarianceThreshold(threshold=0.04384)
# Also trying with some other threshold like lets say 0.03
#selVarianceThreshold2 = VarianceThreshold(threshold=0.03)
Xtrain_removedHalf = selVarianceThreshold2.fit_transform(train[:,:44])
Xtrain_removedHalf.shape
#(55, 22) with 0.04384 and (55, 33) with 0.03
```

```python
# quelles variables ont été conservées ?
print(selVarianceThreshold2.get_support())
```


## <font color='red'> Note
The four random/noise varaibles have not been eliminated, as their variance is higher that the threshold    

Let us see the impact of the new datast with less low variance variables on the performance of the desciion model.


> 

```python
lda = LinearDiscriminantAnalysis()
lda.fit(Xtrain_removedHalf,train[:,44])
lda.score(selVarianceThreshold2.transform(test[:,:44]),test[:,44])
#0.8363636363636363
```

## <font color='red'> Note
The random variables introduced are not eliminated by this method because their variances are among the highest ones. Despite this, we observe an increase in generalisation performance (even compared with the case without random variables), which is probably simply due to a reduction in the complexity of the model : the number of input variables is one of the components of the complexity of a model.

## Univariate feature selection

The second method that we will apply is the univariate feature selection based on [Mutual Information for classification](https://scikit-learn.org/stable/modules/generated/sklearn.feature_selection.mutual_info_classif.html#sklearn.feature_selection.mutual_info_classif). This method eliminates the variables for which the values of the mutual information with the output variable are the lowest (i.e. which 'explain' the output variable least well). 

We intend to keep half the variables and therefore use the `SelectKBest` function (others are available, see the documentation).

> 

```python
from sklearn.feature_selection import SelectKBest, mutual_info_classif
selKBestMutualInfo = SelectKBest(score_func=mutual_info_classif, k=22)
Xtrain_removedHalf = selKBestMutualInfo.fit_transform(train[:,:44], train[:,44])
Xtrain_removedHalf.shape
#(55, 22)
```

```python

# quelles variables ont été conservées (celles avec True) ?
print(selKBestMutualInfo.get_support())
```

## <font color='red'> Note
The first four variables, whose values were random, were eliminated (not selected) by this method.
    
Let us examine the impact on the model performance, and compare with previosu results.. Discuss that !

```python
lda = LinearDiscriminantAnalysis()
lda.fit(Xtrain_removedHalf,train[:,44])
lda.score(selKBestMutualInfo.transform(test[:,:44]),test[:,44])
```

## Sequential feature selection

In this part, we will apply the [Sequential Feature Selection](https://scikit-learn.org/stable/modules/generated/sklearn.feature_selection.SequentialFeatureSelector.html#sklearn.feature_selection.SequentialFeatureSelector) which is a part of *wrapper* methodes. The decision model is same as above [LinearDiscriminantAnalysis](https://scikit-learn.org/stable/modules/generated/sklearn.discriminant_analysis.LinearDiscriminantAnalysis.html)

> 

```python
import sklearn
print('Scikit-learn version:', sklearn.__version__)
## in the terminal, run:  pip install --upgrade scikit-learn
```

```python
##SequentialFeatureSelector, you need sklearn version >= version 0.24.
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis()
selSFS = SequentialFeatureSelector(lda, cv=3, n_features_to_select=22) 
#direction{‘forward’, ‘backward’}, default=’forward’
# cv int, cross-validation generator or an iterable, default=None, Determines the cross-validation splitting strategy.



Xtrain_removedHalf = selSFS.fit_transform(train[:,:44], train[:,44])
Xtrain_removedHalf.shape
#(55, 22)
```

```python

# which variables have not been eliminated (those with True) ?
print(selSFS.get_support())
```

```python
## Let us check the performance of the model
lda = LinearDiscriminantAnalysis()
lda.fit(Xtrain_removedHalf,train[:,44])
lda.score(selSFS.transform(test[:,:44]),test[:,44])
```
